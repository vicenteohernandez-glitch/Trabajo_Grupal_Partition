```python
import time
import random

def partition_base(arr):
    total_sum = sum(arr)
    if total_sum % 2 != 0:
        return False
    target = total_sum // 2

    def backtrack(index, current_sum):
        if current_sum == target: return True
        if current_sum > target or index >= len(arr): return False
        return backtrack(index + 1, current_sum + arr[index]) or backtrack(index + 1, current_sum)

    return backtrack(0, 0)

# --- CONFIGURACIÓN ---
n = 1000 # Cambia este valor manualmente para probar distintos tamaños
random.seed(42) # Semilla fija para replicabilidad
arr = [random.randint(1, 100) for _ in range(n)]

print(f"Ejecutando Fuerza Bruta en Python con n={n}...")

start_time = time.perf_counter()
resultado = partition_base(arr)
end_time = time.perf_counter()

tiempo_ms = (end_time - start_time) * 1000
print(f"Resultado: {resultado}")
print(f"Tiempo de ejecución: {tiempo_ms:.4f} ms")
```

```python
import time
import random

def partition_mejorada(arr):
    total_sum = sum(arr)
    if total_sum % 2 != 0:
        return False

    target = total_sum // 2
    dp = [False] * (target + 1)
    dp[0] = True

    for num in arr:
        for i in range(target, num - 1, -1):
            if dp[i - num]:
                dp[i] = True

    return dp[target]

# --- CONFIGURACIÓN ---
n = 1000  # Cambia este valor manualmente
random.seed(42) # Semilla fija para replicabilidad
arr = [random.randint(1, 100) for _ in range(n)]

print(f"Ejecutando Programación Dinámica en Python con n={n}...")

start_time = time.perf_counter()
resultado = partition_mejorada(arr)
end_time = time.perf_counter()

tiempo_ms = (end_time - start_time) * 1000
print(f"Resultado: {resultado}")
print(f"Tiempo de ejecución: {tiempo_ms:.4f} ms")
```

```python
%%writefile PartitionBruteForce.java
import java.util.Random;

public class PartitionBruteForce {

    private static boolean backtrack(int[] arr, int index, int currentSum, int target) {
        if (currentSum == target) return true;
        if (currentSum > target || index >= arr.length) return false;

        return backtrack(arr, index + 1, currentSum + arr[index], target) ||
               backtrack(arr, index + 1, currentSum, target);
    }

    public static boolean partitionBase(int[] arr) {
        int totalSum = 0;
        for (int num : arr) totalSum += num;
        if (totalSum % 2 != 0) return false;
        return backtrack(arr, 0, 0, totalSum / 2);
    }

    public static void main(String[] args) {
        int n = 1000;

        Random rand = new Random(42);
        int[] arr = new int[n];
        for (int i = 0; i < n; i++) {
            arr[i] = rand.nextInt(100) + 1;
        }

        System.out.println("Ejecutando Fuerza Bruta en Java con n=" + n + "...");

        long startTime = System.nanoTime();
        boolean resultado = partitionBase(arr);
        long endTime = System.nanoTime();

        double tiempo_ms = (endTime - startTime) / 1_000_000.0;

        System.out.println("Resultado: " + resultado);
        System.out.printf("Tiempo de ejecución: %.4f ms\n", tiempo_ms);
    }
}
```

```python
%%writefile PartitionDP.java
import java.util.Random;

public class PartitionDP {

    public static boolean partitionMejorada(int[] arr) {
        int totalSum = 0;
        for (int num : arr) totalSum += num;
        if (totalSum % 2 != 0) return false;

        int target = totalSum / 2;
        boolean[] dp = new boolean[target + 1];
        dp[0] = true;

        for (int num : arr) {
            for (int i = target; i >= num; i--) {
                if (dp[i - num]) {
                    dp[i] = true;
                }
            }
        }
        return dp[target];
    }

    public static void main(String[] args) {
        int n = 1000;

        Random rand = new Random(42);
        int[] arr = new int[n];
        for (int i = 0; i < n; i++) {
            arr[i] = rand.nextInt(100) + 1;
        }

        System.out.println("Ejecutando Programación Dinámica en Java con n=" + n + "...");

        long startTime = System.nanoTime();
        boolean resultado = partitionMejorada(arr);
        long endTime = System.nanoTime();

        double tiempo_ms = (endTime - startTime) / 1_000_000.0;

        System.out.println("Resultado: " + resultado);
        System.out.printf("Tiempo de ejecución: %.4f ms\n", tiempo_ms);
    }
}
```

```python
%%writefile partition_bruteforce.cpp
#include <iostream>
#include <vector>
#include <numeric>
#include <chrono>
#include <random>

using namespace std;

bool backtrack(const vector<int>& arr, int index, int current_sum, int target) {
    if (current_sum == target) return true;
    if (current_sum > target || index >= arr.size()) return false;
    return backtrack(arr, index + 1, current_sum + arr[index], target) ||
           backtrack(arr, index + 1, current_sum, target);
}

bool partition_base(const vector<int>& arr) {
    int total_sum = accumulate(arr.begin(), arr.end(), 0);
    if (total_sum % 2 != 0) return false;
    return backtrack(arr, 0, 0, total_sum / 2);
}

int main() {
    int n = 1000;

    mt19937 gen(42);
    uniform_int_distribution<> dis(1, 100);

    vector<int> arr(n);
    for(int i = 0; i < n; ++i) {
        arr[i] = dis(gen);
    }

    cout << "Ejecutando Fuerza Bruta en C++ con n=" << n << "...\n";

    auto start = chrono::high_resolution_clock::now();
    bool resultado = partition_base(arr);
    auto end = chrono::high_resolution_clock::now();

    chrono::duration<double, std::milli> tiempo_ms = end - start;

    cout << "Resultado: " << (resultado ? "True" : "False") << "\n";
    cout << "Tiempo de ejecucion: " << tiempo_ms.count() << " ms\n";

    return 0;
}
```
